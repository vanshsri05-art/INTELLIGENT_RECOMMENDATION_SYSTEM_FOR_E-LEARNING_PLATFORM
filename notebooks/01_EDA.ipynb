```python
student_ratings = (
    interactions
    .groupby("student_id")["rating"]
    .mean()
    .reset_index()
)

student_ratings.columns = ["student_id", "average_rating"]

student_ratings.head()
```

```python
student_progress = (
    interactions
    .groupby("student_id")["progress"]
    .mean()
    .reset_index()
)

student_progress.columns = ["student_id", "average_progress"]

student_progress.head()
```

```python
completed_courses = (
    interactions[interactions["completed"] == "Yes"]
    .groupby("student_id")
    .size()
    .reset_index(name="courses_completed")
)

completed_courses.head()
```

```python
learning_hours = (
    learning_history
    .groupby("student_id")["hours_spent"]
    .sum()
    .reset_index()
)

learning_hours.columns = ["student_id", "total_learning_hours"]

learning_hours.head()
```

```python
quiz_average = (
    quiz_results
    .groupby("student_id")["quiz_score"]
    .mean()
    .reset_index()
)

quiz_average.columns = ["student_id", "average_quiz_score"]

quiz_average.head()
```

```python
student_features = students[
    ["student_id", "skill_level", "interests"]
].copy()
```

```python
student_features = student_features.merge(
    student_ratings,
    on="student_id",
    how="left"
)

student_features = student_features.merge(
    student_progress,
    on="student_id",
    how="left"
)

student_features = student_features.merge(
    completed_courses,
    on="student_id",
    how="left"
)

student_features = student_features.merge(
    learning_hours,
    on="student_id",
    how="left"
)

student_features = student_features.merge(
    quiz_average,
    on="student_id",
    how="left"
)
```

```python
student_features.head()
```

```python
student_features.isnull().sum()
```

```python
numeric_columns = [
    "average_rating",
    "average_progress",
    "courses_completed",
    "total_learning_hours",
    "average_quiz_score"
]

student_features[numeric_columns] = (
    student_features[numeric_columns].fillna(0)
)
```

```python
student_features.to_csv(
    "../data/student_features.csv",
    index=False
)

print("Student feature dataset saved successfully!")
```
